```python
#Aktivitas 6F

skor = [80, 65, 90, 75, 70]

total = 0                    # akumulator: mulai dari nol
jumlah_lulus = 0             # penghitung = akumulator yang menambah 1
for nilai in skor:
    total = total + nilai    # tambahkan setiap elemen
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

```python
#Aktivitas 6G

suhu = [28, 31, 30, 33, 29, 27, 32]

rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai ->", len(rata_5), "rata-rata")
```

```python
#Aktivitas 6I

suhu = [30, 32, 31, 34, 29, 33, 35]   # 7 suhu harian (bebas)

# (1) 3 hari pertama dan 2 hari terakhir dengan slicing
print("3 hari pertama :", suhu[:3])
print("2 hari terakhir:", suhu[-2:])

# (2) tambah satu hari dengan append, lalu keluarkan lagi dengan pop
suhu.append(31)
print("Setelah append :", suhu)
keluar = suhu.pop()
print("Dikeluarkan    :", keluar, "| panjang:", len(suhu))

# (3) versi urut dengan sorted, suhu asli tidak berubah
urut = sorted(suhu)
print("Urut           :", urut)
print("Asli tetap     :", suhu)

# (4) rata-rata dan suhu tertinggi beserta hari ke berapa
rata = sum(suhu) / len(suhu)
print("Rata-rata      :", round(rata, 1))
i_maks = suhu.index(max(suhu))
print("Tertinggi      :", suhu[i_maks], "pada hari ke-", i_maks + 1)

# (5) list baru berisi suhu di atas rata-rata
di_atas = [t for t in suhu if t > rata]
print("Di atas rata2  :", di_atas)
```

```python
#Aktivitas 6J

nilai = [
    [80, 70, 90],     # baris 0: Ana
    [65, 75, 60],     # baris 1: Budi
    [90, 85, 95],     # baris 2: Cici
]

print(nilai[1])        # satu baris utuh
print(nilai[1][2])     # baris 1, kolom 2: ujian 3 milik Budi
print("baris:", len(nilai), " kolom:", len(nilai[0]))

# Total untuk setiap baris (setiap mahasiswa)
for b in range(len(nilai)):
    print("Mahasiswa", b + 1, "total:", sum(nilai[b]))

# Satu kolom: indeks yang sama di setiap baris (ujian 1 semua orang)
ujian1 = []
for baris in nilai:
    ujian1.append(baris[0])
print("Skor ujian 1:", ujian1)

# 1) Skor ujian 2 milik Cici: baris 2, kolom 1
print("Ujian 2 Cici    :", nilai[2][1])

# 2) Jumlah ujian (jumlah kolom): panjang satu baris
print("Jumlah ujian    :", len(nilai[0]))

# Pembanding: jumlah baris (jumlah mahasiswa)
print("Jumlah mahasiswa:", len(nilai))
```

```python
#Lima Latihan Kecil

skor = [80, 65, 90, 75, 70, 85]

# (1) skor pertama dan terakhir (indeks negatif untuk yang terakhir)
print("Pertama :", skor[0])
print("Terakhir:", skor[-1])

# (2) rata-rata
rata = sum(skor) / len(skor)
print("Rata-rata:", rata)

# (3) berapa skor di atas rata-rata
di_atas = 0
for s in skor:
    if s > rata:
        di_atas += 1
print("Di atas rata-rata:", di_atas)

# (4) posisi skor terendah sebagai nomor mahasiswa (mahasiswa pertama = 1)
i_min = skor.index(min(skor))
print("Skor terendah: mahasiswa ke-", i_min + 1)

# (5) list baru dengan setiap skor ditambah 5
skor_baru = [s + 5 for s in skor]
print("Skor + 5:", skor_baru)
```

```python
#Latihan Praktikum

energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]
n = len(energi)

# (1) Total dan rata-rata
total = sum(energi)
rata = total / n
print("Total     :", total, "MWh")
print("Rata-rata :", round(rata, 1), "MWh/hari")

# (2) Tiga hari tertinggi dan tiga hari terendah (nomor hari, nilai)
pasangan = [(energi[i], i + 1) for i in range(n)]   # (nilai, nomor hari)
urut = sorted(pasangan)
print("\n3 hari tertinggi:")
for nilai, hari in urut[-3:][::-1]:
    print("  hari", hari, "-", nilai, "MWh")
print("3 hari terendah:")
for nilai, hari in urut[:3]:
    print("  hari", hari, "-", nilai, "MWh")

# (3) Jumlah hari di bawah target
target = int(input("\nMasukkan target (MWh): "))
di_bawah = 0
for e in energi:
    if e < target:
        di_bawah += 1
print("Hari di bawah target", target, ":", di_bawah, "dari", n, "hari")

# (4) Rata-rata bergerak 7 hari yang turun di bawah 97% rata-rata bulanan
batas = 0.97 * rata
print("\nBatas 97% rata-rata:", round(batas, 1), "MWh")
ada = False
for i in range(6, n):
    jendela = energi[i - 6 : i + 1]
    rata_7 = sum(jendela) / 7
    if rata_7 < batas:
        print("  jendela hari", i - 5, "-", i + 1, ":", round(rata_7, 1), "MWh")
        ada = True
if not ada:
    print("  tidak ada")

# (5) Total energi per minggu (minggu 1-4, ditambah sisa hari)
print()
for m in range(4):
    mulai = m * 7
    print("Minggu", m + 1, ":", sum(energi[mulai : mulai + 7]), "MWh")
print("Sisa hari (29-30):", sum(energi[28:]), "MWh")
```
