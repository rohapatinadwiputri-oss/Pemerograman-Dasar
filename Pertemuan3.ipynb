1. Urutan prioritas operator (operator precedence)

Berdasarkan dokumentasi resmi Python (docs.python.org, bagian Operator precedence di Language Reference), urutan prioritas dari yang terendah ke yang tertinggi kurang lebih: or → and → not → operator perbandingan (==, <, dll.) → operator bitwise (|, ^, &) → <</>> → +/- → *//////% → unary +x/-x/~x → ** (paling tinggi, kecuali subskrip/pemanggilan fungsi).

Yang relevan untuk soal ini: expresi diselesaikan berdasarkan urutan operator dengan arah tertentu — misalnya perkalian dilakukan sebelum penjumlahan sesuai precedence. Karena operator perkalian (*) berada lebih tinggi dalam tabel prioritas daripada operator penjumlahan (+), Python akan mengevaluasi * terlebih dahulu sebelum +. openssf

Jadi untuk 10 + 5 * 2:

a. Python melihat ada + dan * dalam satu ekspresi.

b. Karena * punya prioritas lebih tinggi, 5 * 2 dihitung lebih dulu → hasilnya 10.

c. Baru kemudian 10 + 10 → hasil akhir 20.

Kalau Python mengevaluasi dari kiri ke kanan tanpa memedulikan prioritas, hasilnya memang akan jadi 30 (yaitu (10 + 5) * 2), tapi itu bukan cara kerja Python — sama seperti aturan matematika biasa, operasi "kali/bagi" selalu didahulukan dari "tambah/kurang" kecuali diberi tanda kurung.

Sumber: Python Software Foundation, "6.17. Operator precedence", The Python Language Reference, docs.python.org/3/reference/expressions.html#operator-precedence.


2. Operator relasional vs operator logika

Operasi relasional vs operator logika
Operasi relasional
- contoh = >, <, ==, !=, >=, <=
- fungsi = Membandingkan dua nilai dan menghasilkan satu True/False
- operand = Angka, string, dll. (nilai apa pun yang bisa dibandingkan)

Operator logic
- contoh = and, or, not
- fungsi = Menggabungkan atau membalik hasil boolean (bisa dari beberapa ekspresi relasional sekaligus)
- operand = Ekspresi/nilai boolean

Jadi operator relasional bekerja di tingkat "membandingkan dua data", sedangkan operator logika bekerja di tingkat "menggabungkan beberapa hasil perbandingan menjadi satu keputusan".

Contoh situasi nyata: Sistem pendaftaran mahasiswa baru mengecek kelayakan pendaftar dengan syarat: umur minimal 17 tahun dan nilai rata-rata rapor minimal 75.

umur = 19

nilai_rapor = 82

if umur >= 17 and nilai_rapor >= 75:
print("diterima")

else :

print("belum memenuhi syarat")

Di sini >= (relasional) dipakai dua kali untuk membandingkan umur dan nilai_rapor masing-masing terhadap batas minimalnya, lalu and (logika) menggabungkan kedua hasil perbandingan itu supaya keputusan "Diterima" hanya muncul kalau kedua syarat terpenuhi sekaligus. Tanpa and, program tidak bisa mengecek dua syarat dalam satu keputusan; tanpa >=, and tidak punya nilai boolean apa pun untuk digabungkan.


3. Contoh kasus nyata bug akibat salah memahami operator precedence

Salah satu contoh nyata yang terdokumentasi adalah kerentanan yang tercatat sebagai CWE-783 (Operator Precedence Logic Error), yaitu kelas kesalahan di mana program menggunakan sebuah ekspresi yang karena precedence operator justru menghasilkan logika yang salah. Salah satu kasus konkretnya adalah CVE-2026-7270 pada penanganan argumen execve di FreeBSD: kode tidak memberi tanda kurung untuk memperjelas urutan operasi yang dimaksud, sehingga pengecekan batas (bounds check) yang seharusnya gagal malah dianggap lolos, dan ini menyebabkan potensi buffer overflow.

Ringkasnya: penulis kode mengira operator-operator dalam ekspresi pengecekan batas buffer akan dievaluasi sesuai urutan "logis" yang mereka bayangkan, padahal precedence bahasa pemrogramannya menggabungkan operand secara berbeda. Karena tidak ada tanda kurung eksplisit, kondisi validasi ukuran data yang dimasukkan pengguna jadi salah dihitung, sehingga data yang seharusnya ditolak (karena melebihi kapasitas buffer) malah diloloskan sistem — pintu masuk bug keamanan.

Seharusnya kode itu diperbaiki dengan menambahkan tanda kurung eksplisit di sekitar sub-ekspresi yang dimaksud, sehingga urutan evaluasi tidak lagi bergantung pada precedence default compiler/interpreter, melainkan sesuai maksud programmer — sama seperti anjuran umum: kalau ekspresi melibatkan lebih dari satu jenis operator dan hasilnya penting (apalagi untuk validasi keamanan), selalu gunakan tanda kurung meskipun secara teknis "tidak wajib", demi kejelasan dan mencegah salah baca urutan operasi.
