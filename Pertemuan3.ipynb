1. Urutan prioritas operator (operator precedence)

Dalam Python terdapat beberapa jenis operator, di antaranya operator aritmatika, operator relasional/perbandingan, operator logika, dan operator assignment (penugasan). Operator aritmatika digunakan untuk melakukan perhitungan seperti +, -, *, /, dan %. Operator relasional digunakan untuk membandingkan nilai, misalnya >, <, ==, dan !=. Operator logika seperti and, or, dan not digunakan untuk menggabungkan atau membalik kondisi, sedangkan assignment seperti = digunakan untuk memberikan nilai kepada variabel.
Jadi untuk 10 + 5 * 2:

Python memiliki aturan operator precedence (prioritas operator), yaitu aturan yang menentukan operator mana yang dikerjakan terlebih dahulu ketika terdapat beberapa operator dalam satu ekspresi. Berdasarkan dokumentasi resmi Python, operator perkalian (*), pembagian (/), floor division (//), dan modulus (%) memiliki prioritas lebih tinggi daripada penjumlahan (+) dan pengurangan (-). Operator yang memiliki prioritas sama umumnya dievaluasi dari kiri ke kanan.

Contohnya :
hasil = 10 + 5 * 4
print(hasil)

Hasilnya adalah :
30

Hal tersebut terjadi karena Python tidak langsung menghitung 10 + 5. Operator * mempunyai prioritas lebih tinggi daripada +, sehingga Python terlebih dahulu menghitung:

5 * 4 = 20

Kemudian hasil tersebut dijumlahkan dengan 10 :

10 + 20 = 30

Jadi, ekspresi:

10 + 5 * 4

Secara sederhana diproses seperti :

10 + (5 * 4)

Bukan :

(10 + 5) * 4

Jika kita memang ingin penjumlahan dilakukan terlebih dahulu, kita harus menggunakan tanda kurung :

(10 + 5) * 4

yang menghasilkan 60. Dokumentasi resmi Python menjelaskan bahwa tanda kurung dapat digunakan untuk mengatur pengelompokan ekspresi dan mengubah urutan pengerjaan sesuai kebutuhan.

2. Operator relasional vs operator logika

Operator relasional atau operator perbandingan digunakan untuk membandingkan dua nilai. Contohnya adalah:
(operator) >, (arti) lebih besar, (contoh) nilai > 70.

(operator) <, (arti) lebih kecil, (contoh) nilai < 70.

(operator) ==, (arti) sama dengan, (contoh) nilai == 70.

(operator) !=, (arti) tidak sama dengan, (contoh) nilai != 70.

(operator) >=, (arti) lebih besar atau sama dengan, (contoh) nilai >= 70.

(operator) <=, (arti) lebih kecil atau sama dengan, (contoh) nilai <= 70

Hasil dari operator perbandingan biasanya berupa nilai True atau False. Dokumentasi Python juga menjelaskan bahwa operasi perbandingan menghasilkan nilai boolean dalam penggunaan normal.

Sedangkan operator logika digunakan untuk menggabungkan atau memanipulasi beberapa kondisi. Operator logika yang umum digunakan adalah :

(operator) and, (fungsi) semua kondisi harus benar.

(operator) or, (fungsi) minimal salah satu kondisi benar.

(operator) not, (fungsi) membalik kondisi.

Contoh situasi nyata :
Misalnya sebuah sistem digunakan untuk menentukan apakah mahasiswa boleh mengikuti ujian. Syaratnya adalah:

- nilai tugas minimal 70, dan
- kehadiran minimal 75%.

Kode Python :

nilai = 80
kehadiran = 85

boleh_ujian = nilai >= 70 and kehadiran >= 75

print(boleh_ujian)

Hasil :

True

Pada contoh tersebut, operator relasional digunakan terlebih dahulu untuk membandingkan nilai :

nilai >= 70

dan kehadiran :

kehadiran >= 75

Keduanya menghasilkan True. Kemudian operator logika and digunakan untuk memastikan kedua kondisi tersebut benar :

True and True

Sehingga hasil akhirnya adalah True.

Contoh ini menunjukkan bahwa operator relasional dan logika memiliki fungsi yang berbeda tetapi dapat digunakan bersama dalam satu ekspresi untuk menghasilkan keputusan. Operator and sendiri hanya akan mengevaluasi kondisi berikutnya jika kondisi sebelumnya bernilai benar.
