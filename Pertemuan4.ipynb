```python
#soal 1

tahun = int(input("masukkan tahun"))
kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
hasil = ["BUKAN TAHUN KABISAT", "TAHUN KABISAT"][kabisat]

print("tahun", tahun, "adalah", hasil)
print()
```

```python
#soal 2
satuan = "cm"
def hitung(panjang, lebar):
  luas = panjang * lebar
  keliling = 2 * (panjang + lebar)
  return luas, keliling

p = float(input("masukkan panjang (" + satuan +"): "))
l = float(input("masukkan lebar (" + satuan +"): "))

luas, keliling = hitung(p, l)
rata_rata_sisi = (p + l) / 2
sisa = int(p) % int(l)

print("luas =", luas, satuan + "²")
print("keliling =", keliling, satuan)
print("rata-rata sisi =", rata_rata_sisi, satuan)
print("sisi bagi panjang/lebar =", sisa)
print()
```
