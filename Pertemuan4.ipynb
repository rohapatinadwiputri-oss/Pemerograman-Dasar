```python
#soal 1

tahun = int(input("masukkan tahun "))
kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
hasil = ["BUKAN TAHUN KABISAT", "TAHUN KABISAT"][kabisat]

print("tahun", tahun, "adalah", hasil)
print()
```

```python
#soal 2
bilangan1 = int(input("masukkan bilangan pertama: "))
bilangan2 = int(input("masukkan bilangan kedua: "))
jumlah = bilangan1 + bilangan2
selisih = bilangan1 - bilangan2
print()

print("hasil penjumlahan =", jumlah)
print("hasil pengurangan =", selisih)
```
